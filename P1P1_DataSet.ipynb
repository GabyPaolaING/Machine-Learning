##PASO 2. Estimación puntual de los parámetros

```python
peso = df_peso['Peso(Kg)']

# Numero de datos
n = len(peso)

print("Número de datos:", n)
```

```python
sumatoria = peso.sum()

print("Sumatoria de los pesos:", round(sumatoria, 2), "kg")
```

```python
media = peso.mean()

print("Media muestral:", round(media, 2), "kg")
```

```python
suma_cuadrados = ((peso - media) ** 2).sum()

print(
    "Suma de cuadrados de las desviaciones:",
    round(suma_cuadrados, 2)
)
```

```python
varianza = peso.var(ddof=1)

print("Varianza muestral:", round(varianza, 2), "kg²")
```

```python
desviacion = peso.std(ddof=1)

print("Desviación estándar:", round(desviacion, 2), "kg")
```

##PASO 3. Error estándar

```python
# PASO 3: ERROR ESTANDAR

error_estandar = desviacion / np.sqrt(n)

print("PASO 3: ERROR ESTANDAR")
print("------------------------------")
print("Desviación estándar:", round(desviacion, 4), "kg")
print("Número de datos:", n)
print("Error estándar:", round(error_estandar, 4), "kg")
```

##PASO 4. Intervalo de confianza

```python
from scipy.stats import t
```

```python
# PASO 4: INTERVALO DE CONFIANZA

confianza = 0.95

# Grados de libertad
grados_libertad = n - 1

# Valor crítico
valor_critico = t.ppf(
    1 - (1 - confianza) / 2,
    grados_libertad
)

# Margen de error
margen_error = valor_critico * error_estandar

# Límites
limite_inferior = media - margen_error
limite_superior = media + margen_error

print("PASO 4: INTERVALO DE CONFIANZA")
print("------------------------------")
print("Nivel de confianza: 95%")
print("Grados de libertad:", grados_libertad)
print("Valor crítico t:", round(valor_critico, 4))
print("Margen de error:", round(margen_error, 4), "kg")
print("Límite inferior:", round(limite_inferior, 4), "kg")
print("Límite superior:", round(limite_superior, 4), "kg")
```

##PASO 5. Prueba de hipótesis

```python
from scipy.stats import ttest_1samp

# PASO 5: PRUEBA DE HIPÓTESIS

umbral = 80.0
alpha = 0.05

# Prueba t de una muestra
estadistico_t, p_valor_dos_colas = ttest_1samp(
    peso,
    umbral
)

# Convertir a prueba unilateral izquierda
if estadistico_t < 0:
    p_valor = p_valor_dos_colas / 2
else:
    p_valor = 1 - (p_valor_dos_colas / 2)

print("PASO 5: PRUEBA DE HIPÓTESIS")
print("------------------------------")
print("H0: A = 80 kg")
print("H1: A < 80 kg")
print("Nivel de significancia:", alpha)
print("Estadístico t:", round(estadistico_t, 4))
print("p-valor unilateral:", round(p_valor, 6))

if p_valor < alpha:
    print("Se rechaza H0.")
else:
    print("No se rechaza H0.")
```

##Paso 6

```python
# PASO 6:

# Crear nuevamente la tabla desde df_limpio
df_peso = df_limpio[['Fecha', 'Peso(Kg)']].copy()

# Convertir Peso a número
df_peso['Peso(Kg)'] = pd.to_numeric(
    df_peso['Peso(Kg)'],
    errors='coerce'
)

# Eliminar el dato incorrecto de 0.3 kg
df_peso = df_peso[df_peso['Peso(Kg)'] != 0.3].copy()

# Convertir Fecha a formato de fecha
df_peso['Fecha'] = pd.to_datetime(
    df_peso['Fecha'],
    errors='coerce'
)

# Tomar solamente mes y día
mes_dia = df_peso['Fecha'].dt.strftime('%m-%d')

# Crear nuevamente las fechas, todas en 2026
df_peso['Fecha'] = pd.to_datetime(
    '2026-' + mes_dia,
    format='%Y-%m-%d',
    errors='coerce'
)

# Ordenar por fecha
df_peso = df_peso.sort_values(
    'Fecha'
).reset_index(drop=True)

# Calcular los días desde la primera medición
df_peso['Dias'] = (
    df_peso['Fecha'] - df_peso['Fecha'].min()
).dt.days

# Mostrar los resultados
display(df_peso[['Fecha', 'Peso(Kg)', 'Dias']])
```

```python
print("Fecha minima:", df_peso['Fecha'].min())
print("Fecha maxima:", df_peso['Fecha'].max())
```

```python
display(df_peso[['Fecha', 'Peso(Kg)', 'Dias']])
```

##Paso 7  

```python
# PASO 7: REGRESIÓN LINEAL

from scipy.stats import linregress

# Variable independiente: días
x = df_peso['Dias']

# Variable dependiente: peso
y = df_peso['Peso(Kg)']

# Crear la regresión lineal
regresion = linregress(x, y)

# Obtener los resultados
pendiente = regresion.slope
intercepto = regresion.intercept
r = regresion.rvalue

# Coeficiente de determinación
r_cuadrada = r ** 2

print("PASO 7: REGRESIÓN LINEAL")
print("------------------------------")
print("Pendiente:", round(pendiente, 6))
print("Intercepto:", round(intercepto, 4))
print("R²:", round(r_cuadrada, 4))

print()
print("Ecuación de la recta:")
print(
    f"Peso estimado = {intercepto:.4f} "
    f"+ ({pendiente:.6f}) × Días"
)
```

##Paso 8

```python
# PASO 8: PESO ESTIMADO

df_peso['Peso_estimado'] = (
    intercepto +
    pendiente * df_peso['Dias']
)

display(df_peso)
```

##Paso 9

```python
# PASO 9: CRUCE DE LA RECTA CON 80 KG

peso_objetivo = 80.0

dia_80 = (
    peso_objetivo - intercepto
) / pendiente

print("PASO 9: CRUCE DE LA RECTA CON 80 KG")
print("------------------------------")
print(
    f"Día t = {dia_80:.2f} días desde la primera medición"
)
```

##Paso 10

```python
# PASO 10: FECHA DEL CRUCE CON 80 KG

fecha_cruce = (
    df_peso['Fecha'].min()
    + pd.to_timedelta(dia_80, unit='D')
)

print("Fecha aproximada del cruce con 80 kg:")
print(fecha_cruce)
```

```python
# PASO 11: GRÁFICA FINAL

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

# 22 datos reales
plt.scatter(
    df_peso['Dias'],
    df_peso['Peso(Kg)'],
    label='Datos reales'
)

# Línea de tendencia
plt.plot(
    df_peso['Dias'],
    df_peso['Peso_estimado'],
    label='Tendencia lineal'
)

# Línea de 80 kg
plt.axhline(
    y=80.0,
    linestyle='--',
    label='Objetivo: 80.0 kg'
)

# Línea de 70.53 kg
plt.axhline(
    y=70.53,
    linestyle='--',
    label='Límite normal: 70.53 kg'
)

# Título
plt.title(
    'Comportamiento del peso a lo largo del tiempo'
)

# Etiquetas
plt.xlabel('Días transcurridos')
plt.ylabel('Peso (kg)')

# Leyenda
plt.legend()

# Cuadrícula
plt.grid(True)

plt.show()
```

##Evaluación Diagnóstica del IMC (Inicial y Actual)

```python
# PASO 1: EVALUACIÓN DEL IMC

estatura_m = 1.68

peso_inicial = df_peso['Peso(Kg)'].iloc[0]
peso_actual = df_peso['Peso(Kg)'].iloc[-1]

# Calcular IMC inicial
imc_inicial = peso_inicial / (estatura_m ** 2)

# Calcular IMC actual
imc_actual = peso_actual / (estatura_m ** 2)

print("EVALUACIÓN DEL IMC")
print("------------------------------")
print("Estatura:", estatura_m, "m")
print("Peso inicial:", peso_inicial, "kg")
print("IMC inicial:", round(imc_inicial, 2))

print()
print("Último peso registrado:", peso_actual, "kg")
print("IMC actual:", round(imc_actual, 2))
```

##Cálculo del Peso Límite Objetivo (Normopeso)


```python
# PASO 2: PESO LÍMITE PARA NORMOPESO

imc_max_normal = 24.9

peso_limite = imc_max_normal * (estatura_m ** 2)

print("PESO LÍMITE PARA NORMOPESO")
print("------------------------------")
print("IMC máximo normal:", imc_max_normal)
print("Estatura:", estatura_m, "m")
print("Peso límite:", round(peso_limite, 2), "kg")
```

##Cuantificación del Remanente (Diferencia de Peso)

```python
# PASO 3: REMANENTE DE PESO

remanente = peso_actual - peso_limite

print("REMANENTE DE PESO")
print("------------------------------")
print("Último peso:", round(peso_actual, 2), "kg")
print("Peso límite:", round(peso_limite, 2), "kg")
print("Diferencia restante:", round(remanente, 2), "kg")
```

##Proyección Temporal con el Modelo de Regresión

```python
# PASO 4: PROYECCIÓN DEL DÍA EN QUE SE ALCANZA EL PESO LÍMITE

dia_normopeso = (
    peso_limite - intercepto
) / pendiente

print("PROYECCIÓN TEMPORAL")
print("------------------------------")
print(
    f"Día estimado para alcanzar {peso_limite:.2f} kg:"
)
print(
    f"{dia_normopeso:.2f} días desde la primera medición"
)
```

```python
# Convertir el día proyectado a una fecha

fecha_normopeso = (
    df_peso['Fecha'].min()
    + pd.to_timedelta(dia_normopeso, unit='D')
)

print("Fecha proyectada:")
print(fecha_normopeso)
```

```python
# PASO 5: GRÁFICA DEL PROGRESO HASTA EL NORMOPESO

import matplotlib.pyplot as plt
import numpy as np

# Crear valores de días desde 0 hasta el día proyectado
dias_proyeccion = np.linspace(
    0,
    dia_normopeso,
    300
)

# Calcular los pesos estimados
pesos_proyectados = (
    intercepto +
    pendiente * dias_proyeccion
)

plt.figure(figsize=(12, 6))

# Datos reales
plt.scatter(
    df_peso['Dias'],
    df_peso['Peso(Kg)'],
    label='Datos reales'
)

# Tendencia y proyección
plt.plot(
    dias_proyeccion,
    pesos_proyectados,
    label='Tendencia y proyección'
)

# Línea del peso límite
plt.axhline(
    y=peso_limite,
    linestyle='--',
    label=f'Límite normopeso: {peso_limite:.2f} kg'
)

# Línea del último peso
plt.axhline(
    y=peso_actual,
    linestyle=':',
    label=f'Último registro: {peso_actual:.2f} kg'
)

# Punto donde se alcanza el peso límite
plt.scatter(
    dia_normopeso,
    peso_limite,
    s=80,
    label='Peso límite proyectado'
)

# Texto del punto
plt.annotate(
    f'{peso_limite:.2f} kg\nDía {dia_normopeso:.2f}',
    (dia_normopeso, peso_limite),
    xytext=(-100, 25),
    textcoords='offset points'
)

# Título
plt.title(
    'Proyección del peso hasta el límite superior del normopeso'
)

# Etiquetas
plt.xlabel('Días desde la primera medición')
plt.ylabel('Peso (kg)')

plt.legend()
plt.grid(True)

plt.show()
```
