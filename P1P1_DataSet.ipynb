```python
# Cargar el archivo
df = pd.read_excel("Datos.xlsx")

# mostrar todos los registros y todas las columnas
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

# para ver el data completo
display(df)
```

##2.- Limpieza y Preprocesamiento

```python

#copea el original y trabaja sobre la version limpia
df_limpio = df.copy()

# Elimina filas vacias
df_limpio = df_limpio.dropna(how='all')

# Elimina columnas vacias
df_limpio = df_limpio.dropna(axis=1, how='all')

df_limpio.columns = df_limpio.iloc[2]

# Elimina las filas anteriores al encabezado
df_limpio = df_limpio.iloc[3:].copy()

# reinicia los indices
df_limpio.reset_index(drop=True, inplace=True)

# muestra la tabla limpia
display(df_limpio)
```

##3.- Análisis Exploratorio de Datos


*  Graficacion

```python
# Convertir la fecha a formato de fecha
df_limpio['Fecha'] = pd.to_datetime(
    df_limpio['Fecha'],
    format='%m-%d',
    errors='coerce'
)

# Ordenar las mediciones de la más antigua a la más reciente
df_limpio = df_limpio.sort_values('Fecha')

# Reiniciar los índices
df_limpio.reset_index(drop=True, inplace=True)
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Convertir el peso a valores numericos
df_limpio['Peso(Kg)'] = pd.to_numeric(df_limpio['Peso(Kg)'], errors='coerce')

# Eliminar datos que no tengan un peso válido
df_grafica = df_limpio.dropna(subset=['Peso(Kg)']).copy()

# Crear la gráfica
plt.figure(figsize=(14, 7))

plt.plot(
    range(1, len(df_grafica) + 1),
    df_grafica['Peso(Kg)'],
    marker='o',
    linewidth=2,
    markersize=5
)

plt.title('Evolución del peso ', fontsize=16)
plt.xlabel('Día de medición', fontsize=12)
plt.ylabel('Peso (Kg)', fontsize=12)

plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(df_limpio['Peso(Kg)'], bins=10)

plt.title('Distribución del peso del sujeto')
plt.xlabel('Peso (Kg)')
plt.ylabel('Frecuencia')
plt.grid(True)

plt.show()
```

*  Calcular la media, mediana y desviación estándar del peso del sujeto a lo largo de los 222 días registrados.  

Media

```python
df_limpio['Peso(Kg)'] = pd.to_numeric(df_limpio['Peso(Kg)'], errors='coerce')

media = df_limpio['Peso(Kg)'].mean()

print("La media del peso es:", media, "Kg")
```

*  Mediana

```python
mediana = df_limpio['Peso(Kg)'].median()

print("La mediana del peso es:", mediana, "Kg")
```

*  Desviacion estandar

```python
desviacion_estandar = df_limpio['Peso(Kg)'].std()

print("La desviación estándar del peso es:", desviacion_estandar, "Kg")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

columnas_relevantes = [
    'Porcentaje de Grasa Corporal',
    'Agua(%)',
    'Musculos(%)',
    'Peso(Kg)'
]

# convertir las columnas a valores numericos
df_correlacion = df_limpio[columnas_relevantes].apply(
    pd.to_numeric,
    errors='coerce'
)

# crear la matriz de correlacion
matriz_correlacion = df_correlacion.corr()

# graficar la matriz de correlacion
plt.figure(figsize=(8, 6))

sns.heatmap(
    matriz_correlacion,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    square=True
)

# titulo de la grafica
plt.title('Matriz de correlación')

# mostrar grafica
plt.show()
```
