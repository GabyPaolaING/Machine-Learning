##PASO 2. Estimación puntual de los parámetros

```python
peso = df_peso['Peso(Kg)']

# Numero de datos
n = len(peso)

print("Número de datos:", n)
```

```python
sumatoria = peso.sum()

print("Sumatoria de los pesos:", round(sumatoria, 2), "kg")
```

```python
media = peso.mean()

print("Media muestral:", round(media, 2), "kg")
```

```python
suma_cuadrados = ((peso - media) ** 2).sum()

print(
    "Suma de cuadrados de las desviaciones:",
    round(suma_cuadrados, 2)
)
```

```python
varianza = peso.var(ddof=1)

print("Varianza muestral:", round(varianza, 2), "kg²")
```

```python
desviacion = peso.std(ddof=1)

print("Desviación estándar:", round(desviacion, 2), "kg")
```

##PASO 3. Error estándar

```python
# PASO 3: ERROR ESTANDAR

error_estandar = desviacion / np.sqrt(n)

print("PASO 3: ERROR ESTANDAR")
print("------------------------------")
print("Desviación estándar:", round(desviacion, 4), "kg")
print("Número de datos:", n)
print("Error estándar:", round(error_estandar, 4), "kg")
```

##PASO 4. Intervalo de confianza

```python
from scipy.stats import t
```

```python
# PASO 4: INTERVALO DE CONFIANZA

confianza = 0.95

# Grados de libertad
grados_libertad = n - 1

# Valor crítico
valor_critico = t.ppf(
    1 - (1 - confianza) / 2,
    grados_libertad
)

# Margen de error
margen_error = valor_critico * error_estandar

# Límites
limite_inferior = media - margen_error
limite_superior = media + margen_error

print("PASO 4: INTERVALO DE CONFIANZA")
print("------------------------------")
print("Nivel de confianza: 95%")
print("Grados de libertad:", grados_libertad)
print("Valor crítico t:", round(valor_critico, 4))
print("Margen de error:", round(margen_error, 4), "kg")
print("Límite inferior:", round(limite_inferior, 4), "kg")
print("Límite superior:", round(limite_superior, 4), "kg")
```

##PASO 5. Prueba de hipótesis

```python
from scipy.stats import ttest_1samp

# PASO 5: PRUEBA DE HIPÓTESIS

umbral = 80.0
alpha = 0.05

# Prueba t de una muestra
estadistico_t, p_valor_dos_colas = ttest_1samp(
    peso,
    umbral
)

# Convertir a prueba unilateral izquierda
if estadistico_t < 0:
    p_valor = p_valor_dos_colas / 2
else:
    p_valor = 1 - (p_valor_dos_colas / 2)

print("PASO 5: PRUEBA DE HIPÓTESIS")
print("------------------------------")
print("H0: A = 80 kg")
print("H1: A < 80 kg")
print("Nivel de significancia:", alpha)
print("Estadístico t:", round(estadistico_t, 4))
print("p-valor unilateral:", round(p_valor, 6))

if p_valor < alpha:
    print("Se rechaza H0.")
else:
    print("No se rechaza H0.")
```

##Paso 6

```python
# PASO 6:

# Crear nuevamente la tabla desde df_limpio
df_peso = df_limpio[['Fecha', 'Peso(Kg)']].copy()

# Convertir Peso a número
df_peso['Peso(Kg)'] = pd.to_numeric(
    df_peso['Peso(Kg)'],
    errors='coerce'
)

# Eliminar el dato incorrecto de 0.3 kg
df_peso = df_peso[df_peso['Peso(Kg)'] != 0.3].copy()

# Convertir Fecha a formato de fecha
df_peso['Fecha'] = pd.to_datetime(
    df_peso['Fecha'],
    errors='coerce'
)

# Tomar solamente mes y día
mes_dia = df_peso['Fecha'].dt.strftime('%m-%d')

# Crear nuevamente las fechas, todas en 2026
df_peso['Fecha'] = pd.to_datetime(
    '2026-' + mes_dia,
    format='%Y-%m-%d',
    errors='coerce'
)

# Ordenar por fecha
df_peso = df_peso.sort_values(
    'Fecha'
).reset_index(drop=True)

# Calcular los días desde la primera medición
df_peso['Dias'] = (
    df_peso['Fecha'] - df_peso['Fecha'].min()
).dt.days

# Mostrar los resultados
display(df_peso[['Fecha', 'Peso(Kg)', 'Dias']])
```

##Paso 7  

```python
# PASO 7: REGRESIÓN LINEAL

from scipy.stats import linregress

# Variable independiente: días
x = df_peso['Dias']

# Variable dependiente: peso
y = df_peso['Peso(Kg)']

# Crear la regresión lineal
regresion = linregress(x, y)

# Obtener los resultados
pendiente = regresion.slope
intercepto = regresion.intercept
r = regresion.rvalue

# Coeficiente de determinación
r_cuadrada = r ** 2

print("PASO 7: REGRESIÓN LINEAL")
print("------------------------------")
print("Pendiente:", round(pendiente, 6))
print("Intercepto:", round(intercepto, 4))
print("R²:", round(r_cuadrada, 4))

print()
print("Ecuación de la recta:")
print(
    f"Peso estimado = {intercepto:.4f} "
    f"+ ({pendiente:.6f}) × Días"
)
```

##Paso 8

```python
# PASO 8: PESO ESTIMADO

df_peso['Peso_estimado'] = (
    intercepto +
    pendiente * df_peso['Dias']
)

display(df_peso)
```

##Paso 9

```python
# PASO 9: CRUCE DE LA RECTA CON 80 KG

peso_objetivo = 80.0

dia_80 = (
    peso_objetivo - intercepto
) / pendiente

print("PASO 9: CRUCE DE LA RECTA CON 80 KG")
print("------------------------------")
print(
    f"Día t = {dia_80:.2f} días desde la primera medición"
)
```

##Paso 10

```python
# PASO 10: FECHA DEL CRUCE CON 80 KG

fecha_cruce = (
    df_peso['Fecha'].min()
    + pd.to_timedelta(dia_80, unit='D')
)

print("Fecha aproximada del cruce con 80 kg:")
print(fecha_cruce)
```

```python
# PASO 2: PESO LÍMITE PARA NORMOPESO

imc_max_normal = 24.9

peso_limite = imc_max_normal * (estatura_m ** 2)

print("PESO LÍMITE PARA NORMOPESO")
print("------------------------------")
print("IMC máximo normal:", imc_max_normal)
print("Estatura:", estatura_m, "m")
print("Peso límite:", round(peso_limite, 2), "kg")
```

```python
# ==========================================
# BLOQUE 3: NORMALIZAR FECHAS DEL EXCEL (CORREGIDO)
# ==========================================

import pandas as pd

# Asegurarnos de limpiar la columna Fecha eliminando espacios si existieran
fechas_limpias = df_peso['Fecha'].astype(str).str.strip()

# Extraer mes y día controlando el formato MM-DD o parseos previos
# Si contiene un año antiguo de pruebas o texto, nos quedamos con los últimos 5 caracteres (MM-DD)
def extraer_mes_dia(val):
    val = str(val).split()[0] # Quitar horas si existen
    if '-' in val:
        partes = val.split('-')
        # Si viene en formato YYYY-MM-DD
        if len(partes[0]) == 4:
            return f"{partes[1]}-{partes[2]}"
        return f"{partes[0]}-{partes[1]}"
    return val

mes_dia_corregido = fechas_limpias.apply(extraer_mes_dia)

# Crear la columna de fecha con el año estándar de la actividad (2026)
df_peso['Fecha_DT'] = pd.to_datetime(
    '2026-' + mes_dia_corregido,
    format='%Y-%m-%d',
    errors='coerce'
)

print("Fechas del Excel normalizadas exitosamente (sin valores NaT):")
display(
    df_peso[
        ['Fecha', 'Fecha_DT', 'Peso(Kg)', 'BMI']
    ].head(10)
)
```

```python

import pandas as pd
import numpy as np
import sqlite3
import json

# ------------------------------------------------------------
# 1. Cargar el Excel usando la fila correcta como encabezado
# ------------------------------------------------------------

df_base = pd.read_excel('Datos.xlsx', header=3)

# Eliminar columnas completamente vacías, si existen
df_base = df_base.dropna(axis=1, how='all')

# Limpiar espacios en los nombres de las columnas
df_base.columns = df_base.columns.astype(str).str.strip()

print("Columnas encontradas:")
print(df_base.columns.tolist())

# ------------------------------------------------------------
# 2. Convertir las fechas al año 2026
# ------------------------------------------------------------

# Las fechas del archivo tienen formato mes-día, por ejemplo 08-16.
# Se indica explícitamente el formato para evitar confusiones.

df_base['Fecha_DT'] = pd.to_datetime(
    '2026-' + df_base['Fecha'].astype(str).str.strip(),
    format='%Y-%m-%d',
    errors='coerce'
)

# Convertir las variables biométricas a números
columnas_numericas = [
    'Peso(Kg)',
    'BMI',
    'Porcentaje de Grasa Corporal',
    'Agua(%)',
    'Musculos(%)'
]

for columna in columnas_numericas:
    if columna in df_base.columns:
        df_base[columna] = pd.to_numeric(
            df_base[columna], errors='coerce'
        )

# Eliminar filas sin fecha válida
df_base = df_base.dropna(subset=['Fecha_DT']).copy()

# Ordenar de la fecha más antigua a la más reciente
df_base = df_base.sort_values('Fecha_DT').reset_index(drop=True)

print("\nPrimeras cinco mediciones:")
display(df_base.head())

print("\nÚltimas cinco mediciones:")
display(df_base.tail())
```

```python

# =====================================================
# BLOQUE 2:
# =====================================================

# -----------------------------------------------------
# 1. Variable binaria: fin de semana
# -----------------------------------------------------

df_base['es_fin_semana'] = (
    df_base['Fecha_DT'].dt.dayofweek >= 5
).astype(int)

# -----------------------------------------------------
# 2. Preparar las sesiones de entrenamiento por fecha
# -----------------------------------------------------

df_sesiones = df_entrenamiento.copy()

df_sesiones['tipo_ejercicio'] = (
    df_sesiones['tipo_ejercicio']
    .fillna('')
    .astype(str)
)

df_sesiones['distancia_km'] = pd.to_numeric(
    df_sesiones['distancia_km'], errors='coerce'
).fillna(0)

df_sesiones['calorias_kcal'] = pd.to_numeric(
    df_sesiones['calorias_kcal'], errors='coerce'
).fillna(0)

# Consideramos ciclismo las sesiones cuyo tipo incluye "ciclismo"
es_ciclismo = df_sesiones['tipo_ejercicio'].str.contains(
    'ciclismo', case=False, na=False
)

# Consideramos fuerza las sesiones de pesas o fuerza
es_fuerza = df_sesiones['tipo_ejercicio'].str.contains(
    'mancuernas|fuerza|pesas', case=False, na=False
)

df_sesiones['km_ciclismo'] = np.where(
    es_ciclismo, df_sesiones['distancia_km'], 0
)

df_sesiones['calorias_fuerza'] = np.where(
    es_fuerza, df_sesiones['calorias_kcal'], 0
)

# Agrupar sesiones de un mismo día
df_diario = df_sesiones.groupby(
    df_sesiones['Fecha_DT'].dt.normalize()
).agg(
    km_ciclismo=('km_ciclismo', 'sum'),
    calorias_fuerza=('calorias_fuerza', 'sum')
)

# Crear calendario diario para no saltarnos días sin entrenamientos
calendario = pd.date_range(
    start=min(df_base['Fecha_DT'].min(), df_diario.index.min()),
    end=max(df_base['Fecha_DT'].max(), df_diario.index.max()),
    freq='D'
)

df_diario = df_diario.reindex(calendario, fill_value=0)
df_diario.index.name = 'Fecha_DT'

# -----------------------------------------------------
# 3. Carga semanal de ciclismo: suma móvil de 7 días
# -----------------------------------------------------

df_diario['carga_semanal_km'] = (
    df_diario['km_ciclismo']
    .rolling(window=7, min_periods=1)
    .sum()
)

# -----------------------------------------------------
# 4. Ratio fuerza/cardio
# -----------------------------------------------------
# Definición: calorías de fuerza de los últimos 7 días
# divididas entre kilómetros de ciclismo de los últimos 7 días.

df_diario['calorias_fuerza_7d'] = (
    df_diario['calorias_fuerza']
    .rolling(window=7, min_periods=1)
    .sum()
)

df_diario['ratio_fuerza_cardio'] = np.where(
    df_diario['carga_semanal_km'] > 0,
    df_diario['calorias_fuerza_7d']
    / df_diario['carga_semanal_km'],
    0
)

# Incorporar variables calculadas al dataset de pesajes
df_base['Fecha_DT_dia'] = df_base['Fecha_DT'].dt.normalize()

df_base = df_base.merge(
    df_diario[
        [
            'carga_semanal_km',
            'ratio_fuerza_cardio'
        ]
    ],
    left_on='Fecha_DT_dia',
    right_index=True,
    how='left'
)

# -----------------------------------------------------
# 5. Distancia al normopeso
# -----------------------------------------------------

peso_objetivo = 70.28

df_base['distancia_al_normopeso'] = (
    df_base['Peso(Kg)'] - peso_objetivo
)

# Un valor positivo indica kg por encima de la meta.
# Un valor negativo indica kg por debajo de la meta.

print("Variables sintéticas creadas:")
display(
    df_base[
        [
            'Fecha_DT',
            'es_fin_semana',
            'carga_semanal_km',
            'ratio_fuerza_cardio',
            'distancia_al_normopeso'
        ]
    ].head(10)
)
```

```python

# =====================================================
# BLOQUE 3: ESTANDARIZACIÓN Z-SCORE
# =====================================================

# Variables biométricas y de actividad que se estandarizarán
columnas_z = [
    'Peso(Kg)',
    'Porcentaje de Grasa Corporal',
    'Agua(%)',
    'Musculos(%)',
    'conteo_pasos',
    'tiempo_actividad_min',
    'calorias_kcal'
]

# Incorporar los datos de actividad según Fecha_DT
# para conservar las columnas de pasos y minutos.
if 'conteo_pasos' not in df_base.columns:
    df_base = df_base.merge(
        df_actividad[
            ['Fecha_DT', 'conteo_pasos', 'tiempo_actividad_min']
        ],
        on='Fecha_DT',
        how='left'
    )

# Incorporar calorías si no se encuentran en el DataFrame.
# Puede haber más de una sesión por fecha; sumamos las calorías.
if 'calorias_kcal' not in df_base.columns:
    calorias_fecha = df_entrenamiento.groupby(
        df_entrenamiento['Fecha_DT'].dt.normalize()
    )['calorias_kcal'].sum()

    df_base['calorias_kcal'] = (
        df_base['Fecha_DT'].dt.normalize()
        .map(calorias_fecha)
    )

# Convertir las variables a números y tratar faltantes.
for columna in columnas_z:
    if columna in df_base.columns:
        df_base[columna] = pd.to_numeric(
            df_base[columna], errors='coerce'
        )

# Los faltantes de pasos y tiempo de actividad representan
# ausencia de registro de actividad, según las instrucciones previas.
for columna in ['conteo_pasos', 'tiempo_actividad_min']:
    if columna in df_base.columns:
        df_base[columna] = df_base[columna].fillna(0)

# Estandarizar con media y desviación estándar muestral.
# Las columnas constantes no se pueden estandarizar: se asigna 0.
for columna in columnas_z:
    if columna in df_base.columns:
        media = df_base[columna].mean()
        desviacion = df_base[columna].std(ddof=1)

        nombre_z = columna.replace('(', '_').replace(')', '').replace('%', 'pct')
        nombre_z = nombre_z.replace(' ', '_') + '_Z'

        if pd.isna(desviacion) or desviacion == 0:
            df_base[nombre_z] = 0
        else:
            df_base[nombre_z] = (
                df_base[columna] - media
            ) / desviacion

# Limpiar columnas auxiliares
df_final = df_base.drop(
    columns=[
        'Cambio_peso',
        'Dias',
        'Peso_estimado',
        'Residuo_peso',
        'Residuo_peso_limpio',
        'Fecha_DT_dia'
    ],
    errors='ignore'
)

# Revisar y tratar los faltantes restantes en columnas numéricas
# y categóricas para obtener un archivo sin valores nulos.
for columna in df_final.select_dtypes(include='number').columns:
    df_final[columna] = df_final[columna].fillna(
        df_final[columna].median()
    )

for columna in df_final.select_dtypes(
    include=['object', 'string']
).columns:
    df_final[columna] = df_final[columna].fillna('Sin registro')

# Exportar el dataset transformado
df_final.to_csv(
    'dataset_salud_rendimiento_transformado.csv',
    index=False,
    encoding='utf-8-sig'
)

print("Tamaño del dataset final:", df_final.shape)
print("Valores nulos restantes:", int(df_final.isna().sum().sum()))
display(df_final.head(10))
```

```python
# ==========================================
# P2P4: MODELADO PREDICTIVO Y SERIES TEMPORALES
# CELDA 1. CARGA DE DATOS
# ==========================================

import pandas as pd
import numpy as np
import sqlite3
import matplotlib.pyplot as plt
import statsmodels.api as sm

from sklearn.linear_model import LinearRegression, RidgeCV, LassoCV
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Cargar archivo Excel
df_peso = pd.read_excel("Datos.xlsx", header=3)

df_peso.columns = df_peso.columns.astype(str).str.strip()

# Convertir las fechas al año 2026
def convertir_fecha(fecha):
    if pd.isna(fecha):
        return pd.NaT

    if isinstance(fecha, (pd.Timestamp, np.datetime64)):
        fecha = pd.Timestamp(fecha)
        return pd.Timestamp(2026, fecha.month, fecha.day)

    try:
        fecha_dt = pd.to_datetime(str(fecha).strip(), format="%m-%d")
        return pd.Timestamp(2026, fecha_dt.month, fecha_dt.day)
    except:
        return pd.to_datetime(fecha, errors="coerce")

df_peso["Fecha_DT"] = df_peso["Fecha"].apply(convertir_fecha)

df_peso["Peso(Kg)"] = pd.to_numeric(
    df_peso["Peso(Kg)"], errors="coerce"
)

df_peso = df_peso.dropna(subset=["Fecha_DT", "Peso(Kg)"])

df_peso = (
    df_peso.sort_values("Fecha_DT")
    .drop_duplicates(subset="Fecha_DT", keep="last")
)

print("Mediciones de peso:", len(df_peso))
display(df_peso[["Fecha_DT", "Peso(Kg)"]].head())
```

```python
# ==========================================
# CELDA 2. ENTRENAMIENTOS DESDE SQLITE
# ==========================================

conexion = sqlite3.connect("salud_ejercicio.db")

df_entrenamiento = pd.read_sql_query(
    "SELECT * FROM sesiones_entrenamiento",
    conexion
)

conexion.close()

df_entrenamiento["fecha"] = pd.to_datetime(
    df_entrenamiento["fecha"], errors="coerce"
).dt.normalize()

df_entrenamiento["distancia_km"] = pd.to_numeric(
    df_entrenamiento["distancia_km"], errors="coerce"
).fillna(0)

df_entrenamiento["calorias_kcal"] = pd.to_numeric(
    df_entrenamiento["calorias_kcal"], errors="coerce"
).fillna(0)

df_entrenamiento["tipo_ejercicio"] = (
    df_entrenamiento["tipo_ejercicio"]
    .astype(str)
    .str.lower()
    .str.strip()
)

# Revisar los nombres de ejercicios
print("Tipos de ejercicio registrados:")
print(df_entrenamiento["tipo_ejercicio"].value_counts())
```

```python
# ==========================================
# CELDA 3. CLASIFICACIÓN DE ENTRENAMIENTOS
# ==========================================

# Identificar MTB o ciclismo al aire libre
es_mtb = df_entrenamiento["tipo_ejercicio"].str.contains(
    r"mtb|montaña|montana|exterior|aire libre|outdoor",
    regex=True,
    na=False
)

# Identificar fuerza o mancuernas
es_fuerza = df_entrenamiento["tipo_ejercicio"].str.contains(
    r"fuerza|mancuernas|pesas",
    regex=True,
    na=False
)

df_entrenamiento["Km_MTB"] = np.where(
    es_mtb,
    df_entrenamiento["distancia_km"],
    0
)

df_entrenamiento["Kcal_Fuerza"] = np.where(
    es_fuerza,
    df_entrenamiento["calorias_kcal"],
    0
)

# Sumar los entrenamientos del mismo día
df_diario = (
    df_entrenamiento.groupby("fecha", as_index=False)
    .agg(
        Km_MTB=("Km_MTB", "sum"),
        Kcal_Fuerza=("Kcal_Fuerza", "sum")
    )
)

display(df_diario.head(10))

print("Total de kilómetros MTB:", df_diario["Km_MTB"].sum())
print("Total de calorías de fuerza:", df_diario["Kcal_Fuerza"].sum())
```

```python

# CELDA 4. Integrar los datos y preparar el modelo predictivo

import pandas as pd

# 1. Preparar las fechas y los datos de entrenamiento
df_entrenamiento["fecha"] = pd.to_datetime(
    df_entrenamiento["fecha"], errors="coerce"
)

df_entrenamiento["Km_MTB"] = pd.to_numeric(
    df_entrenamiento["Km_MTB"], errors="coerce"
).fillna(0)

df_entrenamiento["Kcal_Fuerza"] = pd.to_numeric(
    df_entrenamiento["Kcal_Fuerza"], errors="coerce"
).fillna(0)

# 2. Sumar los kilómetros y las calorías por fecha
actividad_diaria = (
    df_entrenamiento.groupby("fecha")
    .agg(
        Km_MTB=("Km_MTB", "sum"),
        Kcal_Fuerza=("Kcal_Fuerza", "sum")
    )
    .reset_index()
)

# 3. Preparar las mediciones reales de peso
datos = df_peso[["Fecha_DT", "Peso(Kg)"]].copy()

datos["Fecha_DT"] = pd.to_datetime(
    datos["Fecha_DT"], errors="coerce"
)

datos["Peso(Kg)"] = pd.to_numeric(
    datos["Peso(Kg)"], errors="coerce"
)

datos = datos.dropna(subset=["Fecha_DT", "Peso(Kg)"])

# 4. Unir los entrenamientos con las fechas de peso
datos = datos.merge(
    actividad_diaria,
    left_on="Fecha_DT",
    right_on="fecha",
    how="left"
)

# Los ceros indican que no hay entrenamiento registrado
# para esa fecha exacta
datos["Km_MTB"] = datos["Km_MTB"].fillna(0)
datos["Kcal_Fuerza"] = datos["Kcal_Fuerza"].fillna(0)

# 5. Calcular los días transcurridos desde la primera medición
datos["Dias_Transcurridos"] = (
    datos["Fecha_DT"] - datos["Fecha_DT"].min()
).dt.days

# 6. Crear la carga semanal de kilómetros
# Se calcula a partir de las sesiones registradas en los últimos 7 días
calendario = pd.DataFrame({
    "Fecha_DT": pd.date_range(
        start=datos["Fecha_DT"].min(),
        end=datos["Fecha_DT"].max(),
        freq="D"
    )
})

actividad_calendario = calendario.merge(
    actividad_diaria,
    left_on="Fecha_DT",
    right_on="fecha",
    how="left"
)

actividad_calendario["Km_MTB"] = (
    actividad_calendario["Km_MTB"].fillna(0)
)

actividad_calendario["Kcal_Fuerza"] = (
    actividad_calendario["Kcal_Fuerza"].fillna(0)
)

actividad_calendario["Carga_Semanal_Km"] = (
    actividad_calendario["Km_MTB"]
    .rolling(window=7, min_periods=1)
    .sum()
)

# 7. Incorporar la carga semanal a las mediciones de peso
datos = datos.drop(
    columns=["Carga_Semanal_Km"], errors="ignore"
)

datos = datos.merge(
    actividad_calendario[["Fecha_DT", "Carga_Semanal_Km"]],
    on="Fecha_DT",
    how="left"
)

# 8. Preparar las variables independientes y dependiente
X = datos[[
    "Dias_Transcurridos",
    "Carga_Semanal_Km",
    "Kcal_Fuerza"
]].copy()

y = datos["Peso(Kg)"].copy()

# Agregar constante para el modelo OLS
import statsmodels.api as sm

X_ols = sm.add_constant(X, has_constant="add")

# 9. Mostrar resultados para verificar
print("Dimensiones de los datos:", datos.shape)
print("Mediciones de peso:", len(datos))
print("Días con kilómetros registrados:",
      (datos["Carga_Semanal_Km"] > 0).sum())
print("Días con calorías de fuerza registradas:",
      (datos["Kcal_Fuerza"] > 0).sum())

print("\nPrimeras filas de los datos preparados:")
print(datos[[
    "Fecha_DT",
    "Peso(Kg)",
    "Dias_Transcurridos",
    "Carga_Semanal_Km",
    "Kcal_Fuerza"
]].head(10).to_string(index=False))
```

```python
# CELDA 5. Regresión lineal OLS usando el tiempo

import statsmodels.api as sm
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# Variable independiente: días transcurridos
X_tiempo = datos[["Dias_Transcurridos"]]

# Variable dependiente: peso real
y = datos["Peso(Kg)"]

# Agregar constante
X_tiempo_ols = sm.add_constant(X_tiempo)

# Ajustar el modelo
modelo_ols = sm.OLS(y, X_tiempo_ols).fit()

# Mostrar resultados
print(modelo_ols.summary())

# Calcular predicciones
y_pred = modelo_ols.predict(X_tiempo_ols)

# Evaluar el ajuste
mae = mean_absolute_error(y, y_pred)
rmse = np.sqrt(mean_squared_error(y, y_pred))

print("\nMÉTRICAS DEL MODELO OLS")
print(f"MAE: {mae:.3f} kg")
print(f"RMSE: {rmse:.3f} kg")
print(f"R²: {modelo_ols.rsquared:.4f}")

print("\nCOEFICIENTES")
print(modelo_ols.params)
```

```python

# CELDA 6. Gráfica del peso real y el peso estimado

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.scatter(
    datos["Dias_Transcurridos"],
    datos["Peso(Kg)"],
    label="Peso real",
    marker="o"
)

plt.plot(
    datos["Dias_Transcurridos"],
    y_pred,
    label="Peso estimado por OLS",
    linewidth=2
)

plt.axhline(
    y=70.28,
    linestyle="--",
    label="Objetivo: 70.28 kg"
)

plt.xlabel("Días transcurridos")
plt.ylabel("Peso (kg)")
plt.title("Peso real frente al peso estimado mediante OLS")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# CELDA 7. Comparación de modelos OLS, Ridge y Lasso

import numpy as np
import pandas as pd

from sklearn.linear_model import LinearRegression, RidgeCV, LassoCV
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 1. Usar únicamente el tiempo como variable predictora
X_comparacion = datos[["Dias_Transcurridos"]].copy()
y_comparacion = datos["Peso(Kg)"].copy()

# 2. Dividir los datos respetando el orden temporal
punto_corte = int(len(datos) * 0.8)

X_train = X_comparacion.iloc[:punto_corte]
X_test = X_comparacion.iloc[punto_corte:]
y_train = y_comparacion.iloc[:punto_corte]
y_test = y_comparacion.iloc[punto_corte:]

# 3. Crear los modelos
modelo_ols_comp = LinearRegression()

n_splits = min(5, len(X_train) - 1)
cv = TimeSeriesSplit(n_splits=n_splits)

modelo_ridge = make_pipeline(
    StandardScaler(),
    RidgeCV(alphas=[0.01, 0.1, 1, 10, 100], cv=cv)
)

modelo_lasso = make_pipeline(
    StandardScaler(),
    LassoCV(
        alphas=[0.001, 0.01, 0.1, 1, 10],
        cv=cv,
        max_iter=10000
    )
)

# 4. Entrenar los modelos
modelo_ols_comp.fit(X_train, y_train)
modelo_ridge.fit(X_train, y_train)
modelo_lasso.fit(X_train, y_train)

# 5. Obtener predicciones
pred_ols = modelo_ols_comp.predict(X_test)
pred_ridge = modelo_ridge.predict(X_test)
pred_lasso = modelo_lasso.predict(X_test)

# 6. Comparar las métricas
resultados_modelos = []

for nombre, prediccion in [
    ("OLS", pred_ols),
    ("Ridge", pred_ridge),
    ("Lasso", pred_lasso)
]:
    resultados_modelos.append({
        "Modelo": nombre,
        "MAE (kg)": mean_absolute_error(y_test, prediccion),
        "RMSE (kg)": np.sqrt(mean_squared_error(y_test, prediccion)),
        "R²": r2_score(y_test, prediccion)
    })

tabla_comparacion = pd.DataFrame(resultados_modelos)

print("COMPARACIÓN DE MODELOS")
print(tabla_comparacion.round(4).to_string(index=False))
```

```python

# CELDA 8. Comparar las predicciones de los modelos

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.plot(
    datos["Dias_Transcurridos"].iloc[len(X_train):],
    y_test.values,
    marker="o",
    label="Peso real"
)

plt.plot(
    datos["Dias_Transcurridos"].iloc[len(X_train):],
    pred_ols,
    marker="s",
    label="Predicción OLS"
)

plt.plot(
    datos["Dias_Transcurridos"].iloc[len(X_train):],
    pred_ridge,
    marker="^",
    label="Predicción Ridge"
)

plt.plot(
    datos["Dias_Transcurridos"].iloc[len(X_train):],
    pred_lasso,
    marker="x",
    label="Predicción Lasso"
)

plt.axhline(
    y=70.28,
    linestyle="--",
    label="Objetivo: 70.28 kg"
)

plt.xlabel("Días transcurridos")
plt.ylabel("Peso (kg)")
plt.title("Comparación de predicciones: OLS, Ridge y Lasso")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# CELDA 9. Proyección temporal hacia el objetivo

import pandas as pd

objetivo = 70.28

intercepto = modelo_ols.params["const"]
pendiente = modelo_ols.params["Dias_Transcurridos"]

fecha_inicio = datos["Fecha_DT"].min()

print(f"Peso objetivo: {objetivo:.2f} kg")
print(f"Pendiente estimada: {pendiente:.5f} kg por día")

if pendiente < 0:
    dias_estimados = (objetivo - intercepto) / pendiente

    if dias_estimados >= 0:
        fecha_estimada = (
            fecha_inicio + pd.to_timedelta(dias_estimados, unit="D")
        )

        print(f"Días desde el inicio de la serie: {dias_estimados:.0f}")
        print("Fecha estimada:", fecha_estimada.strftime("%d/%m/%Y"))
    else:
        print("El modelo no proyecta alcanzar el objetivo en el futuro.")
else:
    print("La pendiente no indica una disminución del peso.")
    print("No se puede estimar una fecha futura con esta tendencia.")

print("\nADVERTENCIA:")
print("El ajuste temporal es débil (R² aproximado de 0.0083).")
print("La fecha es una extrapolación matemática, no una predicción confiable.")
```

```python

# CELDA 10. Gráfica de la proyección temporal

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

objetivo = 70.28

intercepto = modelo_ols.params["const"]
pendiente = modelo_ols.params["Dias_Transcurridos"]

dias_observados = datos["Dias_Transcurridos"].to_numpy()
pesos_reales = datos["Peso(Kg)"].to_numpy()

plt.figure(figsize=(12, 6))

plt.scatter(
    dias_observados,
    pesos_reales,
    label="Peso real",
    marker="o"
)

# Extender la línea hasta el objetivo, solo si la pendiente baja
if pendiente < 0:
    dias_objetivo = (objetivo - intercepto) / pendiente

    if dias_objetivo > dias_observados.max():
        dias_grafica = np.linspace(
            0, dias_objetivo, 200
        )
    else:
        dias_grafica = np.linspace(
            0, dias_observados.max(), 200
        )
else:
    dias_grafica = np.linspace(
        0, dias_observados.max(), 200
    )

pesos_estimados = intercepto + pendiente * dias_grafica

plt.plot(
    dias_grafica,
    pesos_estimados,
    label="Tendencia estimada por OLS",
    linewidth=2
)

plt.axhline(
    y=objetivo,
    linestyle="--",
    label="Objetivo: 70.28 kg"
)

plt.xlabel("Días transcurridos desde la primera medición")
plt.ylabel("Peso (kg)")
plt.title("Proyección temporal del peso hacia el objetivo")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# CELDA 10: Carga semanal de ciclismo y porcentaje de grasa corporal

# Copiar y preparar los datos
df_relacion = df_peso[
    ["Fecha_DT", "Porcentaje de Grasa Corporal"]
].copy()

# Convertir el porcentaje de grasa a número
df_relacion["Porcentaje de Grasa Corporal"] = pd.to_numeric(
    df_relacion["Porcentaje de Grasa Corporal"],
    errors="coerce"
)

# Preparar las fechas y distancias de entrenamiento
df_bici = df_sesiones_sql.copy()

df_bici["fecha"] = pd.to_datetime(
    df_bici["fecha"], errors="coerce"
)

df_bici["distancia_km"] = pd.to_numeric(
    df_bici["distancia_km"], errors="coerce"
)

# Considerar únicamente ciclismo al aire libre con distancia registrada
df_bici = df_bici[
    df_bici["tipo_ejercicio"].eq("Ciclismo al aire libre")
].copy()

df_bici = df_bici.dropna(
    subset=["fecha", "distancia_km"]
)

# Calcular los kilómetros de los 7 días anteriores
cargas = []

for fecha in df_relacion["Fecha_DT"]:
    inicio = fecha - pd.Timedelta(days=7)

    km_semana = df_bici.loc[
        (df_bici["fecha"] >= inicio) &
        (df_bici["fecha"] < fecha),
        "distancia_km"
    ].sum()

    cargas.append(km_semana)

df_relacion["Carga_Semanal_Km"] = cargas

# Quitar filas sin porcentaje de grasa o fecha
df_relacion = df_relacion.dropna(
    subset=["Fecha_DT", "Porcentaje de Grasa Corporal"]
)

# Mostrar los datos utilizados
display(
    df_relacion[
        ["Fecha_DT", "Carga_Semanal_Km",
         "Porcentaje de Grasa Corporal"]
    ]
)

# Crear la gráfica
plt.figure(figsize=(10, 6))

sns.regplot(
    data=df_relacion,
    x="Carga_Semanal_Km",
    y="Porcentaje de Grasa Corporal",
    scatter_kws={"s": 70},
    line_kws={"color": "red"}
)

plt.title(
    "Carga semanal de ciclismo y porcentaje de grasa corporal"
)
plt.xlabel("Carga semanal de ciclismo (km)")
plt.ylabel("Grasa corporal (%)")
plt.tight_layout()
plt.show()
```

```python

# CELDA 13: Dashboard final de rendimiento y composición corporal

# Preparar los datos de peso
df_dashboard = df_peso.dropna(
    subset=["Fecha_DT", "Peso(Kg)"]
).copy()

df_dashboard = df_dashboard[
    df_dashboard["Peso(Kg)"] > 0
].sort_values("Fecha_DT")

df_dashboard["Dias"] = (
    df_dashboard["Fecha_DT"] -
    df_dashboard["Fecha_DT"].min()
).dt.days

pendiente, intercepto, r, p, error = linregress(
    df_dashboard["Dias"],
    df_dashboard["Peso(Kg)"]
)

df_dashboard["Tendencia"] = (
    pendiente * df_dashboard["Dias"] + intercepto
)

# Preparar los datos de Xiaomi
df_xiaomi = pd.DataFrame(datos_xiaomi).copy()
df_xiaomi["fecha"] = pd.to_datetime(
    df_xiaomi["fecha"], errors="coerce"
)

# Preparar los datos de sesiones
df_sesiones = df_sesiones_sql.copy()
df_sesiones["fecha"] = pd.to_datetime(
    df_sesiones["fecha"], errors="coerce"
)

# Contar las sesiones por tipo
sesiones_deporte = df_sesiones["tipo_ejercicio"].value_counts()

# Crear la figura con cuatro cuadrantes
fig, axes = plt.subplots(2, 2, figsize=(18, 12))
fig.suptitle(
    "Dashboard de Rendimiento y Composición Corporal",
    fontsize=18,
    fontweight="bold"
)

# 1. Evolución del peso
ax = axes[0, 0]

ax.scatter(
    df_dashboard["Fecha_DT"],
    df_dashboard["Peso(Kg)"],
    label="Peso registrado"
)

ax.plot(
    df_dashboard["Fecha_DT"],
    df_dashboard["Tendencia"],
    label="Línea de regresión"
)

ax.axhline(
    70.28,
    color="red",
    linestyle=":",
    label="Objetivo: 70.28 kg"
)

ax.set_title("1. Evolución del peso corporal")
ax.set_xlabel("Fecha")
ax.set_ylabel("Peso (kg)")
ax.legend()
ax.tick_params(axis="x", rotation=45)

# 2. Dona: distribución de sesiones
ax = axes[0, 1]

ax.pie(
    sesiones_deporte.values,
    labels=sesiones_deporte.index,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width": 0.40}
)

ax.set_title("2. Distribución de sesiones por ejercicio")

# 3. Carga semanal frente a grasa corporal
ax = axes[1, 0]

sns.regplot(
    data=df_relacion_limpio,
    x="Carga_Semanal_Km",
    y="Porcentaje de Grasa Corporal",
    ax=ax,
    scatter_kws={"s": 50},
    line_kws={"color": "red"}
)

ax.set_title("3. Carga semanal vs. grasa corporal")
ax.set_xlabel("Carga semanal de ciclismo (km)")
ax.set_ylabel("Grasa corporal (%)")

# 4. Pasos diarios de Xiaomi y sesiones de entrenamiento
ax = axes[1, 1]

pasos_por_dia = df_xiaomi.groupby(
    df_xiaomi["fecha"].dt.date
)["conteo_pasos"].sum()

sesiones_por_dia = df_sesiones.groupby(
    df_sesiones["fecha"].dt.date
).size()

fechas = sorted(
    set(pasos_por_dia.index) |
    set(sesiones_por_dia.index)
)

pasos = [
    pasos_por_dia.get(fecha, 0)
    for fecha in fechas
]

sesiones = [
    sesiones_por_dia.get(fecha, 0)
    for fecha in fechas
]

ax.bar(
    fechas,
    pasos,
    color="steelblue",
    label="Pasos Xiaomi"
)

ax.set_title("4. Actividad diaria y entrenamientos")
ax.set_xlabel("Fecha")
ax.set_ylabel("Pasos")
ax.tick_params(axis="x", rotation=45)

# Eje secundario para contar sesiones
ax2 = ax.twinx()

ax2.plot(
    fechas,
    sesiones,
    color="darkorange",
    marker="o",
    label="Sesiones"
)

ax2.set_ylabel("Sesiones de entrenamiento")

# Ajustar espacios
fig.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```
