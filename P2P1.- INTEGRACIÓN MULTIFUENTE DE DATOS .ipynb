##PASO 2. Estimación puntual de los parámetros

```python
peso = df_peso['Peso(Kg)']

# Numero de datos
n = len(peso)

print("Número de datos:", n)
```

```python
sumatoria = peso.sum()

print("Sumatoria de los pesos:", round(sumatoria, 2), "kg")
```

```python
media = peso.mean()

print("Media muestral:", round(media, 2), "kg")
```

```python
suma_cuadrados = ((peso - media) ** 2).sum()

print(
    "Suma de cuadrados de las desviaciones:",
    round(suma_cuadrados, 2)
)
```

```python
varianza = peso.var(ddof=1)

print("Varianza muestral:", round(varianza, 2), "kg²")
```

```python
desviacion = peso.std(ddof=1)

print("Desviación estándar:", round(desviacion, 2), "kg")
```

##PASO 3. Error estándar

```python
# PASO 3: ERROR ESTANDAR

error_estandar = desviacion / np.sqrt(n)

print("PASO 3: ERROR ESTANDAR")
print("------------------------------")
print("Desviación estándar:", round(desviacion, 4), "kg")
print("Número de datos:", n)
print("Error estándar:", round(error_estandar, 4), "kg")
```

##PASO 4. Intervalo de confianza

```python
from scipy.stats import t
```

```python
# PASO 4: INTERVALO DE CONFIANZA

confianza = 0.95

# Grados de libertad
grados_libertad = n - 1

# Valor crítico
valor_critico = t.ppf(
    1 - (1 - confianza) / 2,
    grados_libertad
)

# Margen de error
margen_error = valor_critico * error_estandar

# Límites
limite_inferior = media - margen_error
limite_superior = media + margen_error

print("PASO 4: INTERVALO DE CONFIANZA")
print("------------------------------")
print("Nivel de confianza: 95%")
print("Grados de libertad:", grados_libertad)
print("Valor crítico t:", round(valor_critico, 4))
print("Margen de error:", round(margen_error, 4), "kg")
print("Límite inferior:", round(limite_inferior, 4), "kg")
print("Límite superior:", round(limite_superior, 4), "kg")
```

##PASO 5. Prueba de hipótesis

```python
from scipy.stats import ttest_1samp

# PASO 5: PRUEBA DE HIPÓTESIS

umbral = 80.0
alpha = 0.05

# Prueba t de una muestra
estadistico_t, p_valor_dos_colas = ttest_1samp(
    peso,
    umbral
)

# Convertir a prueba unilateral izquierda
if estadistico_t < 0:
    p_valor = p_valor_dos_colas / 2
else:
    p_valor = 1 - (p_valor_dos_colas / 2)

print("PASO 5: PRUEBA DE HIPÓTESIS")
print("------------------------------")
print("H0: A = 80 kg")
print("H1: A < 80 kg")
print("Nivel de significancia:", alpha)
print("Estadístico t:", round(estadistico_t, 4))
print("p-valor unilateral:", round(p_valor, 6))

if p_valor < alpha:
    print("Se rechaza H0.")
else:
    print("No se rechaza H0.")
```

##Paso 6

```python
# PASO 6:

# Crear nuevamente la tabla desde df_limpio
df_peso = df_limpio[['Fecha', 'Peso(Kg)']].copy()

# Convertir Peso a número
df_peso['Peso(Kg)'] = pd.to_numeric(
    df_peso['Peso(Kg)'],
    errors='coerce'
)

# Eliminar el dato incorrecto de 0.3 kg
df_peso = df_peso[df_peso['Peso(Kg)'] != 0.3].copy()

# Convertir Fecha a formato de fecha
df_peso['Fecha'] = pd.to_datetime(
    df_peso['Fecha'],
    errors='coerce'
)

# Tomar solamente mes y día
mes_dia = df_peso['Fecha'].dt.strftime('%m-%d')

# Crear nuevamente las fechas, todas en 2026
df_peso['Fecha'] = pd.to_datetime(
    '2026-' + mes_dia,
    format='%Y-%m-%d',
    errors='coerce'
)

# Ordenar por fecha
df_peso = df_peso.sort_values(
    'Fecha'
).reset_index(drop=True)

# Calcular los días desde la primera medición
df_peso['Dias'] = (
    df_peso['Fecha'] - df_peso['Fecha'].min()
).dt.days

# Mostrar los resultados
display(df_peso[['Fecha', 'Peso(Kg)', 'Dias']])
```

##Paso 7  

```python
# PASO 7: REGRESIÓN LINEAL

from scipy.stats import linregress

# Variable independiente: días
x = df_peso['Dias']

# Variable dependiente: peso
y = df_peso['Peso(Kg)']

# Crear la regresión lineal
regresion = linregress(x, y)

# Obtener los resultados
pendiente = regresion.slope
intercepto = regresion.intercept
r = regresion.rvalue

# Coeficiente de determinación
r_cuadrada = r ** 2

print("PASO 7: REGRESIÓN LINEAL")
print("------------------------------")
print("Pendiente:", round(pendiente, 6))
print("Intercepto:", round(intercepto, 4))
print("R²:", round(r_cuadrada, 4))

print()
print("Ecuación de la recta:")
print(
    f"Peso estimado = {intercepto:.4f} "
    f"+ ({pendiente:.6f}) × Días"
)
```

##Paso 8

```python
# PASO 8: PESO ESTIMADO

df_peso['Peso_estimado'] = (
    intercepto +
    pendiente * df_peso['Dias']
)

display(df_peso)
```

##Paso 9

```python
# PASO 9: CRUCE DE LA RECTA CON 80 KG

peso_objetivo = 80.0

dia_80 = (
    peso_objetivo - intercepto
) / pendiente

print("PASO 9: CRUCE DE LA RECTA CON 80 KG")
print("------------------------------")
print(
    f"Día t = {dia_80:.2f} días desde la primera medición"
)
```

##Paso 10

```python
# PASO 10: FECHA DEL CRUCE CON 80 KG

fecha_cruce = (
    df_peso['Fecha'].min()
    + pd.to_timedelta(dia_80, unit='D')
)

print("Fecha aproximada del cruce con 80 kg:")
print(fecha_cruce)
```

```python
# PASO 1: EVALUACIÓN DEL IMC

estatura_m = 1.68

peso_inicial = df_peso['Peso(Kg)'].iloc[0]
peso_actual = df_peso['Peso(Kg)'].iloc[-1]

# Calcular IMC inicial
imc_inicial = peso_inicial / (estatura_m ** 2)

# Calcular IMC actual
imc_actual = peso_actual / (estatura_m ** 2)

print("EVALUACIÓN DEL IMC")
print("------------------------------")
print("Estatura:", estatura_m, "m")
print("Peso inicial:", peso_inicial, "kg")
print("IMC inicial:", round(imc_inicial, 2))

print()
print("Último peso registrado:", peso_actual, "kg")
print("IMC actual:", round(imc_actual, 2))
```

##Cálculo del Peso Límite Objetivo (Normopeso)


```python
# PASO 2: PESO LÍMITE PARA NORMOPESO

imc_max_normal = 24.9

peso_limite = imc_max_normal * (estatura_m ** 2)

print("PESO LÍMITE PARA NORMOPESO")
print("------------------------------")
print("IMC máximo normal:", imc_max_normal)
print("Estatura:", estatura_m, "m")
print("Peso límite:", round(peso_limite, 2), "kg")
```

##Cuantificación del Remanente (Diferencia de Peso)

```python
# PASO 3: REMANENTE DE PESO

remanente = peso_actual - peso_limite

print("REMANENTE DE PESO")
print("------------------------------")
print("Último peso:", round(peso_actual, 2), "kg")
print("Peso límite:", round(peso_limite, 2), "kg")
print("Diferencia restante:", round(remanente, 2), "kg")
```

```python
# ==========================================
# BLOQUE 3: NORMALIZAR FECHAS DEL EXCEL (CORREGIDO)
# ==========================================

import pandas as pd

# Asegurarnos de limpiar la columna Fecha eliminando espacios si existieran
fechas_limpias = df_peso['Fecha'].astype(str).str.strip()

# Extraer mes y día controlando el formato MM-DD o parseos previos
# Si contiene un año antiguo de pruebas o texto, nos quedamos con los últimos 5 caracteres (MM-DD)
def extraer_mes_dia(val):
    val = str(val).split()[0] # Quitar horas si existen
    if '-' in val:
        partes = val.split('-')
        # Si viene en formato YYYY-MM-DD
        if len(partes[0]) == 4:
            return f"{partes[1]}-{partes[2]}"
        return f"{partes[0]}-{partes[1]}"
    return val

mes_dia_corregido = fechas_limpias.apply(extraer_mes_dia)

# Crear la columna de fecha con el año estándar de la actividad (2026)
df_peso['Fecha_DT'] = pd.to_datetime(
    '2026-' + mes_dia_corregido,
    format='%Y-%m-%d',
    errors='coerce'
)

print("Fechas del Excel normalizadas exitosamente (sin valores NaT):")
display(
    df_peso[
        ['Fecha', 'Fecha_DT', 'Peso(Kg)', 'BMI']
    ].head(10)
)
```

```python

import pandas as pd
import numpy as np
import sqlite3
import json

# ------------------------------------------------------------
# 1. Cargar el Excel usando la fila correcta como encabezado
# ------------------------------------------------------------

df_base = pd.read_excel('Datos.xlsx', header=3)

# Eliminar columnas completamente vacías, si existen
df_base = df_base.dropna(axis=1, how='all')

# Limpiar espacios en los nombres de las columnas
df_base.columns = df_base.columns.astype(str).str.strip()

print("Columnas encontradas:")
print(df_base.columns.tolist())

# ------------------------------------------------------------
# 2. Convertir las fechas al año 2026
# ------------------------------------------------------------

# Las fechas del archivo tienen formato mes-día, por ejemplo 08-16.
# Se indica explícitamente el formato para evitar confusiones.

df_base['Fecha_DT'] = pd.to_datetime(
    '2026-' + df_base['Fecha'].astype(str).str.strip(),
    format='%Y-%m-%d',
    errors='coerce'
)

# Convertir las variables biométricas a números
columnas_numericas = [
    'Peso(Kg)',
    'BMI',
    'Porcentaje de Grasa Corporal',
    'Agua(%)',
    'Musculos(%)'
]

for columna in columnas_numericas:
    if columna in df_base.columns:
        df_base[columna] = pd.to_numeric(
            df_base[columna], errors='coerce'
        )

# Eliminar filas sin fecha válida
df_base = df_base.dropna(subset=['Fecha_DT']).copy()

# Ordenar de la fecha más antigua a la más reciente
df_base = df_base.sort_values('Fecha_DT').reset_index(drop=True)

print("\nPrimeras cinco mediciones:")
display(df_base.head())

print("\nÚltimas cinco mediciones:")
display(df_base.tail())
```

```python

# =====================================================
# BLOQUE 2:
# =====================================================

# -----------------------------------------------------
# 1. Variable binaria: fin de semana
# -----------------------------------------------------

df_base['es_fin_semana'] = (
    df_base['Fecha_DT'].dt.dayofweek >= 5
).astype(int)

# -----------------------------------------------------
# 2. Preparar las sesiones de entrenamiento por fecha
# -----------------------------------------------------

df_sesiones = df_entrenamiento.copy()

df_sesiones['tipo_ejercicio'] = (
    df_sesiones['tipo_ejercicio']
    .fillna('')
    .astype(str)
)

df_sesiones['distancia_km'] = pd.to_numeric(
    df_sesiones['distancia_km'], errors='coerce'
).fillna(0)

df_sesiones['calorias_kcal'] = pd.to_numeric(
    df_sesiones['calorias_kcal'], errors='coerce'
).fillna(0)

# Consideramos ciclismo las sesiones cuyo tipo incluye "ciclismo"
es_ciclismo = df_sesiones['tipo_ejercicio'].str.contains(
    'ciclismo', case=False, na=False
)

# Consideramos fuerza las sesiones de pesas o fuerza
es_fuerza = df_sesiones['tipo_ejercicio'].str.contains(
    'mancuernas|fuerza|pesas', case=False, na=False
)

df_sesiones['km_ciclismo'] = np.where(
    es_ciclismo, df_sesiones['distancia_km'], 0
)

df_sesiones['calorias_fuerza'] = np.where(
    es_fuerza, df_sesiones['calorias_kcal'], 0
)

# Agrupar sesiones de un mismo día
df_diario = df_sesiones.groupby(
    df_sesiones['Fecha_DT'].dt.normalize()
).agg(
    km_ciclismo=('km_ciclismo', 'sum'),
    calorias_fuerza=('calorias_fuerza', 'sum')
)

# Crear calendario diario para no saltarnos días sin entrenamientos
calendario = pd.date_range(
    start=min(df_base['Fecha_DT'].min(), df_diario.index.min()),
    end=max(df_base['Fecha_DT'].max(), df_diario.index.max()),
    freq='D'
)

df_diario = df_diario.reindex(calendario, fill_value=0)
df_diario.index.name = 'Fecha_DT'

# -----------------------------------------------------
# 3. Carga semanal de ciclismo: suma móvil de 7 días
# -----------------------------------------------------

df_diario['carga_semanal_km'] = (
    df_diario['km_ciclismo']
    .rolling(window=7, min_periods=1)
    .sum()
)

# -----------------------------------------------------
# 4. Ratio fuerza/cardio
# -----------------------------------------------------
# Definición: calorías de fuerza de los últimos 7 días
# divididas entre kilómetros de ciclismo de los últimos 7 días.

df_diario['calorias_fuerza_7d'] = (
    df_diario['calorias_fuerza']
    .rolling(window=7, min_periods=1)
    .sum()
)

df_diario['ratio_fuerza_cardio'] = np.where(
    df_diario['carga_semanal_km'] > 0,
    df_diario['calorias_fuerza_7d']
    / df_diario['carga_semanal_km'],
    0
)

# Incorporar variables calculadas al dataset de pesajes
df_base['Fecha_DT_dia'] = df_base['Fecha_DT'].dt.normalize()

df_base = df_base.merge(
    df_diario[
        [
            'carga_semanal_km',
            'ratio_fuerza_cardio'
        ]
    ],
    left_on='Fecha_DT_dia',
    right_index=True,
    how='left'
)

# -----------------------------------------------------
# 5. Distancia al normopeso
# -----------------------------------------------------

peso_objetivo = 70.28

df_base['distancia_al_normopeso'] = (
    df_base['Peso(Kg)'] - peso_objetivo
)

# Un valor positivo indica kg por encima de la meta.
# Un valor negativo indica kg por debajo de la meta.

print("Variables sintéticas creadas:")
display(
    df_base[
        [
            'Fecha_DT',
            'es_fin_semana',
            'carga_semanal_km',
            'ratio_fuerza_cardio',
            'distancia_al_normopeso'
        ]
    ].head(10)
)
```

```python

# =====================================================
# BLOQUE 3: ESTANDARIZACIÓN Z-SCORE
# =====================================================

# Variables biométricas y de actividad que se estandarizarán
columnas_z = [
    'Peso(Kg)',
    'Porcentaje de Grasa Corporal',
    'Agua(%)',
    'Musculos(%)',
    'conteo_pasos',
    'tiempo_actividad_min',
    'calorias_kcal'
]

# Incorporar los datos de actividad según Fecha_DT
# para conservar las columnas de pasos y minutos.
if 'conteo_pasos' not in df_base.columns:
    df_base = df_base.merge(
        df_actividad[
            ['Fecha_DT', 'conteo_pasos', 'tiempo_actividad_min']
        ],
        on='Fecha_DT',
        how='left'
    )

# Incorporar calorías si no se encuentran en el DataFrame.
# Puede haber más de una sesión por fecha; sumamos las calorías.
if 'calorias_kcal' not in df_base.columns:
    calorias_fecha = df_entrenamiento.groupby(
        df_entrenamiento['Fecha_DT'].dt.normalize()
    )['calorias_kcal'].sum()

    df_base['calorias_kcal'] = (
        df_base['Fecha_DT'].dt.normalize()
        .map(calorias_fecha)
    )

# Convertir las variables a números y tratar faltantes.
for columna in columnas_z:
    if columna in df_base.columns:
        df_base[columna] = pd.to_numeric(
            df_base[columna], errors='coerce'
        )

# Los faltantes de pasos y tiempo de actividad representan
# ausencia de registro de actividad, según las instrucciones previas.
for columna in ['conteo_pasos', 'tiempo_actividad_min']:
    if columna in df_base.columns:
        df_base[columna] = df_base[columna].fillna(0)

# Estandarizar con media y desviación estándar muestral.
# Las columnas constantes no se pueden estandarizar: se asigna 0.
for columna in columnas_z:
    if columna in df_base.columns:
        media = df_base[columna].mean()
        desviacion = df_base[columna].std(ddof=1)

        nombre_z = columna.replace('(', '_').replace(')', '').replace('%', 'pct')
        nombre_z = nombre_z.replace(' ', '_') + '_Z'

        if pd.isna(desviacion) or desviacion == 0:
            df_base[nombre_z] = 0
        else:
            df_base[nombre_z] = (
                df_base[columna] - media
            ) / desviacion

# Limpiar columnas auxiliares
df_final = df_base.drop(
    columns=[
        'Cambio_peso',
        'Dias',
        'Peso_estimado',
        'Residuo_peso',
        'Residuo_peso_limpio',
        'Fecha_DT_dia'
    ],
    errors='ignore'
)

# Revisar y tratar los faltantes restantes en columnas numéricas
# y categóricas para obtener un archivo sin valores nulos.
for columna in df_final.select_dtypes(include='number').columns:
    df_final[columna] = df_final[columna].fillna(
        df_final[columna].median()
    )

for columna in df_final.select_dtypes(
    include=['object', 'string']
).columns:
    df_final[columna] = df_final[columna].fillna('Sin registro')

# Exportar el dataset transformado
df_final.to_csv(
    'dataset_salud_rendimiento_transformado.csv',
    index=False,
    encoding='utf-8-sig'
)

print("Tamaño del dataset final:", df_final.shape)
print("Valores nulos restantes:", int(df_final.isna().sum().sum()))
display(df_final.head(10))
```

```python

# ============================================================
# BLOQUE 1. INTEGRAR DATOS Y CALCULAR CORRELACIONES
# ============================================================

import pandas as pd
import numpy as np
import sqlite3
import json
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# ------------------------------------------------------------
# 1. Cargar correctamente el Excel
# ------------------------------------------------------------

df_base = pd.read_excel('Datos.xlsx', header=3)
df_base = df_base.dropna(axis=1, how='all')
df_base.columns = df_base.columns.astype(str).str.strip()

# Convertir fechas MM-DD al año 2026
fecha_texto = df_base['Fecha'].astype(str).str.strip()
fecha_parseada = pd.to_datetime(
    fecha_texto, format='%m-%d', errors='coerce'
)

df_base['Fecha_DT'] = pd.to_datetime({
    'year': 2026,
    'month': fecha_parseada.dt.month,
    'day': fecha_parseada.dt.day
}, errors='coerce')

df_base = df_base.dropna(subset=['Fecha_DT']).copy()

# Convertir variables biométricas a números
for col in [
    'Peso(Kg)', 'Porcentaje de Grasa Corporal',
    'Musculos(%)', 'Agua(%)', 'BMI'
]:
    if col in df_base.columns:
        df_base[col] = pd.to_numeric(df_base[col], errors='coerce')

# ------------------------------------------------------------
# 2. Cargar actividad desde el archivo JSON
# ------------------------------------------------------------

with open('pasos_actividad_xiaomi.json', 'r', encoding='utf-8') as archivo:
    datos_json = json.load(archivo)

df_actividad = pd.DataFrame(datos_json)

df_actividad['Fecha_DT'] = pd.to_datetime(
    df_actividad['fecha'], errors='coerce'
)

df_actividad = df_actividad[
    ['Fecha_DT', 'conteo_pasos', 'tiempo_actividad_min']
].copy()

# ------------------------------------------------------------
# 3. Cargar entrenamientos desde SQLite
# ------------------------------------------------------------

with sqlite3.connect('salud_ejercicio.db') as conexion:
    df_entrenamiento = pd.read_sql_query(
        """
        SELECT fecha, tipo_ejercicio, distancia_km, calorias_kcal
        FROM sesiones_entrenamiento
        """,
        conexion
    )

df_entrenamiento['Fecha_DT'] = pd.to_datetime(
    df_entrenamiento['fecha'], errors='coerce'
)

for col in ['distancia_km', 'calorias_kcal']:
    df_entrenamiento[col] = pd.to_numeric(
        df_entrenamiento[col], errors='coerce'
    )

# Sumar los registros de entrenamiento de cada día
df_entrenamiento = df_entrenamiento.groupby(
    'Fecha_DT', as_index=False
).agg(
    tipo_ejercicio=(
        'tipo_ejercicio',
        lambda x: ', '.join(x.dropna().astype(str))
    ),
    distancia_km=('distancia_km', lambda x: x.sum(min_count=1)),
    calorias_kcal=('calorias_kcal', lambda x: x.sum(min_count=1))
)

# ------------------------------------------------------------
# 4. Crear un calendario diario común para las tres fuentes
# ------------------------------------------------------------

fechas_validas = pd.concat([
    df_base['Fecha_DT'],
    df_actividad['Fecha_DT'],
    df_entrenamiento['Fecha_DT']
]).dropna()

calendario = pd.DataFrame({
    'Fecha_DT': pd.date_range(
        fechas_validas.min().normalize(),
        fechas_validas.max().normalize(),
        freq='D'
    )
})

# Integrar las fuentes por fecha
df = calendario.merge(df_base, on='Fecha_DT', how='left')
df = df.merge(df_actividad, on='Fecha_DT', how='left')
df = df.merge(df_entrenamiento, on='Fecha_DT', how='left')

# Los registros de entrenamiento inexistentes se interpretan
# como cero actividad registrada ese día.
df['distancia_km'] = df['distancia_km'].fillna(0)
df['calorias_kcal'] = df['calorias_kcal'].fillna(0)
df['tipo_ejercicio'] = df['tipo_ejercicio'].fillna('Sin registro')

# No rellenar pasos faltantes fuera del periodo cubierto por el JSON.
# Así no se confunde la falta de información con cero pasos.

# ------------------------------------------------------------
# 5. Calcular la carga semanal de ciclismo en kilómetros
# ------------------------------------------------------------

# Sumar las distancias de cada día y acumular los últimos 7 días
df = df.sort_values('Fecha_DT').reset_index(drop=True)

df['Carga_Semanal_Km'] = (
    df.set_index('Fecha_DT')['distancia_km']
      .rolling('7D', min_periods=1)
      .sum()
      .to_numpy()
)

# ------------------------------------------------------------
# 6. Calcular las matrices de correlación
# ------------------------------------------------------------

columnas_corr = [
    'Peso(Kg)',
    'Porcentaje de Grasa Corporal',
    'Musculos(%)',
    'Carga_Semanal_Km',
    'conteo_pasos'
]

for col in columnas_corr:
    df[col] = pd.to_numeric(df[col], errors='coerce')

print("Cantidad de valores disponibles por variable:")
display(df[columnas_corr].count().to_frame('Datos disponibles'))

pearson = df[columnas_corr].corr(method='pearson')
spearman = df[columnas_corr].corr(method='spearman')

print("Correlación de Pearson:")
display(pearson.round(3))

print("Correlación de Spearman:")
display(spearman.round(3))

# Graficar las matrices
fig, ejes = plt.subplots(1, 2, figsize=(16, 6))

sns.heatmap(
    pearson, annot=True, cmap='coolwarm',
    vmin=-1, vmax=1, fmt='.2f', ax=ejes[0]
)
ejes[0].set_title('Correlación de Pearson')

sns.heatmap(
    spearman, annot=True, cmap='coolwarm',
    vmin=-1, vmax=1, fmt='.2f', ax=ejes[1]
)
ejes[1].set_title('Correlación de Spearman')

plt.tight_layout()
plt.show()

print("Dimensiones del conjunto integrado:", df.shape)
```

```python

# ============================================================
# BLOQUE 2. ANALISIS DE ACTIVIDAD POR DIA DE LA SEMANA
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# ------------------------------------------------------------
# 1. Preparar los datos de entrenamiento
# ------------------------------------------------------------

# Usamos df_entrenamiento, creado en el bloque 1
actividad_dias = df_entrenamiento.copy()

actividad_dias['Fecha_DT'] = pd.to_datetime(
    actividad_dias['Fecha_DT'], errors='coerce'
)

actividad_dias['distancia_km'] = pd.to_numeric(
    actividad_dias['distancia_km'], errors='coerce'
).fillna(0)

actividad_dias['calorias_kcal'] = pd.to_numeric(
    actividad_dias['calorias_kcal'], errors='coerce'
).fillna(0)

actividad_dias = actividad_dias.dropna(
    subset=['Fecha_DT']
).copy()

# ------------------------------------------------------------
# 2. Identificar el día de la semana
# ------------------------------------------------------------

actividad_dias['Dia_ingles'] = (
    actividad_dias['Fecha_DT'].dt.day_name()
)

nombres_dias = {
    'Monday': 'Lunes',
    'Tuesday': 'Martes',
    'Wednesday': 'Miercoles',
    'Thursday': 'Jueves',
    'Friday': 'Viernes',
    'Saturday': 'Sabado',
    'Sunday': 'Domingo'
}

actividad_dias['Dia_semana'] = (
    actividad_dias['Dia_ingles'].map(nombres_dias)
)

# ------------------------------------------------------------
# 3. Clasificar los tipos de entrenamiento
# ------------------------------------------------------------

def clasificar_ejercicio(tipo):
    tipo = str(tipo).lower()

    if 'interior' in tipo:
        return 'Ciclismo de interior'
    elif 'aire libre' in tipo:
        return 'Ciclismo al aire libre'
    elif 'mancuerna' in tipo or 'fuerza' in tipo:
        return 'Fuerza y mancuernas'
    else:
        return 'Otro ejercicio'

actividad_dias['Categoria'] = (
    actividad_dias['tipo_ejercicio'].apply(
        clasificar_ejercicio
    )
)

# Clasificar días laborables y fines de semana
actividad_dias['Tipo_dia'] = np.where(
    actividad_dias['Fecha_DT'].dt.dayofweek >= 5,
    'Fin de semana',
    'Dia laborable'
)

# ------------------------------------------------------------
# 4. Crear tabla de actividad por día de la semana
# ------------------------------------------------------------

orden_dias = [
    'Lunes', 'Martes', 'Miercoles', 'Jueves',
    'Viernes', 'Sabado', 'Domingo'
]

tabla_dias = actividad_dias.groupby(
    'Dia_semana'
).agg(
    Numero_sesiones=('tipo_ejercicio', 'count'),
    Distancia_total_km=('distancia_km', 'sum'),
    Distancia_promedio_km=('distancia_km', 'mean'),
    Calorias_totales=('calorias_kcal', 'sum')
)

tabla_dias = tabla_dias.reindex(orden_dias)

print("RESUMEN DE ACTIVIDAD POR DIA DE LA SEMANA")
display(tabla_dias.round(2))

# ------------------------------------------------------------
# 5. Comparar el tipo de entrenamiento por día
# ------------------------------------------------------------

tabla_tipos = pd.crosstab(
    actividad_dias['Dia_semana'],
    actividad_dias['Categoria']
).reindex(orden_dias, fill_value=0)

print("SESIONES POR DIA Y TIPO DE EJERCICIO")
display(tabla_tipos)

# ------------------------------------------------------------
# 6. Comparar días laborables y fines de semana
# ------------------------------------------------------------

resumen_tipo_dia = actividad_dias.groupby(
    'Tipo_dia'
).agg(
    Numero_sesiones=('tipo_ejercicio', 'count'),
    Distancia_promedio_km=('distancia_km', 'mean'),
    Calorias_promedio=('calorias_kcal', 'mean')
)

print("COMPARACION ENTRE DIAS LABORABLES Y FINES DE SEMANA")
display(resumen_tipo_dia.round(2))

# ------------------------------------------------------------
# 7. Graficar la distancia recorrida por día
# ------------------------------------------------------------

tabla_dias['Distancia_total_km'].plot(
    kind='bar',
    figsize=(10, 5)
)

plt.title('Distancia total recorrida por dia de la semana')
plt.xlabel('Dia de la semana')
plt.ylabel('Distancia total (km)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 8. Graficar las sesiones por tipo de entrenamiento
# ------------------------------------------------------------

tabla_tipos.plot(
    kind='bar',
    figsize=(12, 6)
)

plt.title('Tipos de entrenamiento por dia de la semana')
plt.xlabel('Dia de la semana')
plt.ylabel('Numero de sesiones')
plt.xticks(rotation=45)
plt.legend(title='Tipo de ejercicio')
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 9. Comparación estadística exploratoria de distancia
# ------------------------------------------------------------

dist_laborables = actividad_dias.loc[
    (actividad_dias['Tipo_dia'] == 'Dia laborable')
    & (actividad_dias['distancia_km'] > 0),
    'distancia_km'
]

dist_fin_semana = actividad_dias.loc[
    (actividad_dias['Tipo_dia'] == 'Fin de semana')
    & (actividad_dias['distancia_km'] > 0),
    'distancia_km'
]

print("Sesiones con distancia en días laborables:",
      len(dist_laborables))
print("Sesiones con distancia en fines de semana:",
      len(dist_fin_semana))

if len(dist_laborables) >= 2 and len(dist_fin_semana) >= 2:

    prueba = stats.ttest_ind(
        dist_laborables,
        dist_fin_semana,
        equal_var=False
    )

    print("Prueba t de Welch para distancia:")
    print("Estadistico t:", round(prueba.statistic, 4))
    print("Valor p:", round(prueba.pvalue, 4))

    if prueba.pvalue < 0.05:
        print(
            "Existe evidencia de una diferencia estadisticamente "
            "significativa en la distancia media."
        )
    else:
        print(
            "No se encontro evidencia suficiente de una diferencia "
            "estadisticamente significativa en la distancia media."
        )

else:
    print(
        "No hay suficientes sesiones con distancia registrada "
        "en ambos grupos para realizar la prueba t."
    )
```

```python

# ============================================================
# BLOQUE 3. PRUEBA T PARA PÉRDIDA DE PESO SEMANAL
# ============================================================

# Preparar las mediciones de peso
pesos = df[['Fecha_DT', 'Peso(Kg)']].copy()

pesos['Fecha_DT'] = pd.to_datetime(
    pesos['Fecha_DT'], errors='coerce'
)

pesos['Peso(Kg)'] = pd.to_numeric(
    pesos['Peso(Kg)'], errors='coerce'
)

pesos = pesos.dropna(
    subset=['Fecha_DT', 'Peso(Kg)']
).sort_values('Fecha_DT')

# Calcular el peso inicial y final de cada semana
pesos['Semana'] = (
    pesos['Fecha_DT'].dt.to_period('W-SUN')
)

resumen_peso = pesos.groupby('Semana').agg(
    Peso_inicial=('Peso(Kg)', 'first'),
    Peso_final=('Peso(Kg)', 'last'),
    Mediciones=('Peso(Kg)', 'count')
)

# Solo comparar semanas con al menos dos mediciones
resumen_peso = resumen_peso[
    resumen_peso['Mediciones'] >= 2
].copy()

resumen_peso['Perdida_peso_kg'] = (
    resumen_peso['Peso_inicial']
    - resumen_peso['Peso_final']
)

# Calcular kilómetros por semana usando los entrenamientos
entrenos = df_entrenamiento.copy()

entrenos['Fecha_DT'] = pd.to_datetime(
    entrenos['Fecha_DT'], errors='coerce'
)

entrenos['distancia_km'] = pd.to_numeric(
    entrenos['distancia_km'], errors='coerce'
).fillna(0)

entrenos = entrenos.dropna(subset=['Fecha_DT'])

entrenos['Semana'] = (
    entrenos['Fecha_DT'].dt.to_period('W-SUN')
)

km_semana = entrenos.groupby('Semana')['distancia_km'].sum()

resumen_peso['Km_semana'] = (
    km_semana.reindex(resumen_peso.index).fillna(0)
)

# Clasificar las semanas
alto_volumen = resumen_peso.loc[
    resumen_peso['Km_semana'] > 80,
    'Perdida_peso_kg'
].dropna()

bajo_volumen = resumen_peso.loc[
    resumen_peso['Km_semana'] < 40,
    'Perdida_peso_kg'
].dropna()

print("Semanas de alto volumen:", len(alto_volumen))
print("Semanas de bajo volumen:", len(bajo_volumen))

print("\nResumen de las semanas comparadas:")
display(resumen_peso.round(2))

# Ejecutar la prueba t solo si existen suficientes semanas
if len(alto_volumen) >= 2 and len(bajo_volumen) >= 2:

    # Welch: no exige que las varianzas sean iguales
    resultado_t = stats.ttest_ind(
        alto_volumen,
        bajo_volumen,
        equal_var=False
    )

    print("\nPromedio de pérdida en alto volumen:",
          round(alto_volumen.mean(), 3), "kg")

    print("Promedio de pérdida en bajo volumen:",
          round(bajo_volumen.mean(), 3), "kg")

    print("Estadístico t:", round(resultado_t.statistic, 4))
    print("Valor p:", round(resultado_t.pvalue, 4))

    if resultado_t.pvalue < 0.05:
        print(
            "Conclusión: se observa una diferencia estadísticamente "
            "significativa entre los grupos (p < 0.05)."
        )
    else:
        print(
            "Conclusión: no se encontró evidencia suficiente de una "
            "diferencia estadísticamente significativa (p >= 0.05)."
        )

else:
    print(
        "\nNo hay suficientes semanas con mediciones de peso "
        "en ambos grupos para realizar una prueba t confiable."
    )
    print(
        "Se necesitan al menos dos semanas en cada grupo; "
        "es preferible contar con más observaciones."
    )
```
